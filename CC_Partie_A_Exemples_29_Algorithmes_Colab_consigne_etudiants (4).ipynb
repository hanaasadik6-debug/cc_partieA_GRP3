Nom et famille : Régression linéaire / Linear Regression, appartenant à la famille de la régression supervisée.   Objectif : Ajuster une droite ou un hyperplan qui minimise les erreurs quadratiques entre les valeurs cibles et les prédictions.   Intuition et étapes : Le modèle prend en entrée une matrice de caractéristiques $X$ et un vecteur cible $y$. Il calcule une combinaison linéaire des entrées pour prédire la sortie, puis optimise ses coefficients par la méthode des moindres ordinaires pour minimiser l'erreur globale.   Formule et symboles :$$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij}, \qquad J = \frac{1}{2n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$Définition des symboles : $n$ = nombre de lignes ; $p$ = nombre de caractéristiques ; $x_{ij}$ = variable $j$ de l'observation $i$ ; $y_i$ = vraie valeur cible ; $\beta_0$ = constante (biais) ; $\beta_j$ = coefficient estimé ; $J$ = coût moyen quadratique.   Paramètres et hyperparamètres : Les paramètres appris sont les coefficients $\beta_0$ et $\beta_j$. L'hyperparamètre principal est fit_intercept (booléen pour inclure ou non la constante).   Avantages et limites :Avantages : Modèle très rapide, hautement explicable et interprétable.   Limites : Très sensible aux valeurs atypiques (outliers) et incapable de modéliser des relations non linéaires de base.   Schéma explicatifDescription du schéma : Un nuage de points en deux dimensions représentant des observations $(x, y)$, traversé par une droite ajustée, avec des segments verticaux reliant chaque point à la droite.Comment lire ce schéma : Les segments verticaux matérialisent les résidus (les erreurs $y_i - \hat{y}_i$). L'algorithme cherche à positionner la droite de manière à minimiser la somme des carrés de ces segments verticaux.Contextes d'utilisationCas d'usage concrets :Finance / Gestion : Prévision du chiffre d'affaires d'une entreprise en fonction des dépenses publicitaires.   Logistique : Estimation du coût d'acheminement d'une marchandise selon son poids et la distance à parcourir.Nature des données et prérequis : Variables explicatives essentiellement numériques, relation sous-jacente linéaire, absence de colinéarité excessive entre les variables.Cas peu approprié : Données présentant des relations fortement non linéaires ou comportant de nombreux bruits et valeurs aberrantes.Métriques adaptées : MAE (Erreur Absolue Moyenne), RMSE (Racine de l'Erreur Quadratique Moyenne) et $R^2$ (pouvoir descriptif). Hyperparamètres : Utilisation des paramètres par défaut de LinearRegression() (fit_intercept=True).   Sortie obtenue : La valeur de la MAE calculée sur le jeu de test.   Conclusion : La sortie permet de quantifier l'erreur moyenne de prédiction sur le jeu de test synthétique. Cependant, elle ne permet pas d'établir de lien de causalité direct entre les variables, ni de garantir de la même performance sur des données économiques réelles.   Modèle type pour la Partie A : 8. Régression logistiquePrésentationNom et famille : Régression logistique / Logistic Regression, appartenant à la famille de la classification supervisée.   Objectif : Transformer un score linéaire en une probabilité d'appartenance à une classe à l'aide d'une fonction sigmoïde.   Intuition et étapes : Le modèle calcule une combinaison linéaire des caractéristiques (le logit), la convertit en une probabilité comprise entre 0 et 1 via la fonction logistique, puis applique un seuil de décision pour classifier l'observation.   Formule et symboles :$$p(x) = \sigma(z) = \frac{1}{1 + e^{-z}}, \quad z = \beta_0 + \beta^\top x$$Définition des symboles : $p(x)$ = probabilité de la classe 1 ; $\sigma$ = fonction sigmoïde ; $z$ = score linéaire (logit) ; $\beta_0$ = biais ; $\beta$ = vecteur des poids ; $x$ = caractéristiques.   Paramètres et hyperparamètres : Les coefficients $\beta_0$ et $\beta$ sont appris. L'hyperparamètre principal est $C$ (l'inverse de la force de régularisation).   Avantages et limites :Avantages : Fournit des probabilités directement interprétables et un coût de calcul faible.   Limites : Ne peut tracer que des frontières de décision linéaires dans l'espace d'origine, sauf si on lui fournit des transformations de variables.   Schéma explicatifDescription du schéma : Une courbe en forme de "S" (la sigmoïde) tracée sur un graphique, avec un axe des abscisses représentant le score linéaire et un axe des ordonnées représentant la probabilité de 0 à 1, ainsi qu'une ligne de seuil à 0,5.Comment lire ce schéma : Tout score supérieur au seuil (généralement 0,5) positionne l'observation du côté de la classe positive (1), tandis qu'un score inférieur la classe du côté de la classe négative (0).Contextes d'utilisationCas d'usage concrets :Finance : Prédiction du défaut de paiement d'un client (scoring de crédit).   Marketing : Prédire si un client va souscrire ou non à une offre promotionnelle après une campagne.Nature des données et prérequis : Cible binaire (0 ou 1), caractéristiques numériques idéalement standardisées pour stabiliser l'optimisation.Cas peu approprié : Problèmes de classification dont les frontières entre classes sont extrêmement complexes et non linéaires.Métriques adaptées : F1-score, Précision, Rappel, ROC-AUC et PR-AUC (particulièrement si les classes sont déséquilibrées).
````python
# Données pédagogiques reproductibles (les étiquettes sont connues pour l'évaluation)
import sys, platform, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from sklearn.datasets import make_regression, make_classification, make_blobs, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score, accuracy_score,
                             f1_score, precision_score, recall_score, roc_auc_score,
                             average_precision_score, silhouette_score, adjusted_rand_score)
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
warnings.filterwarnings('ignore', category=FutureWarning)

GRAINE = 42
# --- Données pédagogiques SYNTHÉTIQUES (étiquettes connues pour l'évaluation) ---
X_reg, y_reg = make_regression(n_samples=280, n_features=5, noise=13, random_state=GRAINE)
X_cls, y_cls = make_classification(n_samples=400, n_features=6, n_informative=4, n_redundant=0, random_state=GRAINE)
X_clu, etiquettes_reference = make_blobs(n_samples=240, centers=3, cluster_std=0.85, random_state=GRAINE)
Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
# Sous-ensemble de VALIDATION tiré de l'entraînement : sert à comparer des hyperparamètres sans toucher au test
Xr_sub, Xr_val, yr_sub, yr_val = train_test_split(Xr_ent, yr_ent, test_size=.25, random_state=GRAINE)
Xc_sub, Xc_val, yc_sub, yc_val = train_test_split(Xc_ent, yc_ent, test_size=.25, stratify=yc_ent, random_state=GRAINE)

pd.set_option('display.width', 200); pd.set_option('display.max_columns', 20)
RESULTATS_REG, RESULTATS_CLS = {}, {}

def rapport_reg(modele, nom):
    """MAE, RMSE, R² sur le jeu de TEST ; range le résultat dans RESULTATS_REG."""
    p = modele.predict(Xr_test)
    r = {'MAE': mean_absolute_error(yr_test, p), 'RMSE': float(np.sqrt(mean_squared_error(yr_test, p))), 'R2': r2_score(yr_test, p)}
    RESULTATS_REG[nom] = r
    print(f"{nom} | MAE={r['MAE']:.3f} RMSE={r['RMSE']:.3f} R²={r['R2']:.3f}")
    return r

def rapport_cls(modele, nom):
    """F1, précision, rappel, ROC-AUC, PR-AUC sur le jeu de TEST ; range le résultat dans RESULTATS_CLS."""
    p = modele.predict(Xc_test)
    if hasattr(modele, 'predict_proba'):
        s = modele.predict_proba(Xc_test)[:, 1]
    else:
        s = modele.decision_function(Xc_test)
    r = {'F1': f1_score(yc_test, p), 'Précision': precision_score(yc_test, p), 'Rappel': recall_score(yc_test, p),
         'ROC-AUC': roc_auc_score(yc_test, s), 'PR-AUC': average_precision_score(yc_test, s)}
    RESULTATS_CLS[nom] = r
    print(f"{nom} | " + ' '.join(f'{k}={v:.3f}' for k, v in r.items()))
    return r

print('Python', platform.python_version(), '| numpy', np.__version__, '| pandas', pd.__version__, '| scikit-learn', sklearn.__version__)
print('Régression:', X_reg.shape, '| Classification:', X_cls.shape, '| Clustering:', X_clu.shape)
````

````python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
rapport_reg(modele, '1. Régression linéaire')
print('Constante:', round(float(modele.intercept_), 3), '| Coefficients:', np.round(modele.coef_, 2))
````

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score

# Création d'un jeu de validation local à partir des données d'entraînement
Xr_sub, Xr_val, yr_sub, yr_val = train_test_split(Xr_ent, yr_ent, test_size=0.25, random_state=GRAINE)

for degre in [1, 2, 3]:
    m = make_pipeline(StandardScaler(), PolynomialFeatures(degree=degre, include_bias=False), Ridge(alpha=10)).fit(Xr_sub, yr_sub)
    print('degré =', degre, '| MAE validation =', round(mean_absolute_error(yr_val, m.predict(Xr_val)), 3))

# Modèle final entraîné sur l'ensemble de train
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=10)).fit(Xr_ent, yr_ent)

# Évaluation finale sur le jeu de test
y_pred = modele.predict(Xr_test)
print('R² test (degré 2):', r2_score(yr_test, y_pred))
print('MAE test (degré 2):', mean_absolute_error(yr_test, y_pred))
print('Nombre de variables après expansion (degré 2):', modele[1].n_output_features_)
````

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
vraies_probas = np.array([.2, .5, .7])
def bandit_uniforme(graine, T=500):
    g = np.random.default_rng(graine); somme = np.zeros(3); comptes = np.zeros(3); cumul = 0
    for t in range(T):
        bras = t % 3
        gain = int(g.random() < vraies_probas[bras]); somme[bras] += gain; comptes[bras] += 1; cumul += gain
    return somme / comptes, cumul
estimees, cumul = bandit_uniforme(GRAINE)
print('Valeurs estimées:', np.round(estimees, 3), '| Récompense cumulée:', cumul)
print('Récompense cumulée espérée du meilleur bras seul (500 × 0,7):', 500 * vraies_probas.max())
cumuls = [bandit_uniforme(s)[1] for s in range(20)]
print('Allocation uniforme sur 20 graines : moyenne =', round(float(np.mean(cumuls)), 1), '| écart-type =', round(float(np.std(cumuls)), 1))
````

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
vraies_probas = np.array([.2, .5, .7])
def eps_greedy(epsilon, graine, T=1000):
    g = np.random.default_rng(graine); somme = np.zeros(3); comptes = np.zeros(3); cumul = 0
    for t in range(T):
        if g.random() < epsilon:
            bras = int(g.integers(3))
        else:
            bras = int(np.argmax(somme / np.maximum(comptes, 1)))   # moyenne simple Q_t(a), comme dans le cours
        gain = int(g.random() < vraies_probas[bras]); somme[bras] += gain; comptes[bras] += 1; cumul += gain
    return comptes, somme / np.maximum(comptes, 1), cumul
comptes, valeurs, cumul = eps_greedy(.1, GRAINE)
print('Epsilon=0,1 (graine 42) | Nombre de choix:', comptes, '| Valeurs estimées:', np.round(valeurs, 3), '| Récompense cumulée:', cumul)
for eps in [0, .1, .3]:
    cumuls = [eps_greedy(eps, s)[2] for s in range(20)]
    print(f'epsilon={eps:<4} | récompense cumulée sur 20 graines : moyenne={np.mean(cumuls):.1f} écart-type={np.std(cumuls):.1f}')
print('Rappel : meilleur bras seul = 1000 × 0,7 = 700 ; allocation uniforme ≈ 1000 × 0,467 ≈ 467')
````

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import gymnasium as gym
RL = {}
def creer_env():
    return gym.make('FrozenLake-v1', is_slippery=False)

def entrainer_q(graine, episodes=4000, alpha=.3, gamma=.95):
    env = creer_env(); Q = np.zeros((env.observation_space.n, env.action_space.n)); generateur = np.random.default_rng(graine); gains = []
    for episode in range(episodes):
        etat, _ = env.reset(seed=graine + episode); total = 0
        for pas in range(100):
            action = int(generateur.integers(env.action_space.n)) if generateur.random() < max(.05, 1 - episode / 2500) else int(np.argmax(Q[etat]))
            suivant, recompense, termine, tronque, _ = env.step(action)
            cible = recompense + (0 if termine else gamma * np.max(Q[suivant]))
            Q[etat, action] += alpha * (cible - Q[etat, action]); etat = suivant; total += recompense
            if termine or tronque: break
        gains.append(total)
    env.close(); return Q, np.array(gains)

def evaluer_politique(Q, episodes=100):
    """Taux de réussite de la politique gloutonne (argmax Q) : l'agent n'explore plus."""
    env = creer_env(); succes = 0
    for e in range(episodes):
        etat, _ = env.reset(seed=10_000 + e)
        for pas in range(100):
            etat, recompense, termine, tronque, _ = env.step(int(np.argmax(Q[etat])))
            if termine or tronque: break
        succes += int(recompense > 0)
    env.close(); return succes / episodes

def taux_aleatoire(episodes=500, graine=GRAINE):
    env = creer_env(); g = np.random.default_rng(graine); succes = 0
    for e in range(episodes):
        etat, _ = env.reset(seed=20_000 + e)
        for pas in range(100):
            etat, recompense, termine, tronque, _ = env.step(int(g.integers(env.action_space.n)))
            if termine or tronque: break
        succes += int(recompense > 0)
    env.close(); return succes / episodes

Q, gains = entrainer_q(GRAINE)
print('Politique Q-learning (argmax par état):', Q.argmax(axis=1))
fleches = np.where(Q.max(axis=1) > 0, np.array(['←', '↓', '→', '↑'])[Q.argmax(axis=1)], '·')
print('Grille 4×4 (· = état sans valeur apprise) :\n', fleches.reshape(4, 4))
print('Taux de réussite (politique gloutonne, graine 42):', evaluer_politique(Q))
print('Taux de réussite d\'une politique aléatoire:', taux_aleatoire())
taux = [evaluer_politique(entrainer_q(s)[0]) for s in range(5)]
RL['Q-learning'] = taux
print('Q-learning sur 5 graines : taux de réussite =', taux, '| moyenne =', round(float(np.mean(taux)), 3))
plt.plot(np.convolve(gains, np.ones(100) / 100, mode='valid')); plt.xlabel('Épisode'); plt.ylabel('Récompense moyenne (fenêtre 100)'); plt.title('Q-learning : apprentissage sur FrozenLake'); plt.show()
````

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
def entrainer_sarsa(graine, episodes=4000, alpha=.3, gamma=.95):
    env = creer_env(); nA = env.action_space.n; Q = np.zeros((env.observation_space.n, nA)); generateur = np.random.default_rng(graine); gains = []
    def choisir(etat, epsilon):
        return int(generateur.integers(nA)) if generateur.random() < epsilon else int(np.argmax(Q[etat]))
    for episode in range(episodes):
        etat, _ = env.reset(seed=graine + episode); epsilon = max(.05, 1 - episode / 2500); action = choisir(etat, epsilon); total = 0
        for pas in range(100):
            suivant, recompense, termine, tronque, _ = env.step(action)
            action_suiv = choisir(suivant, epsilon)
            cible = recompense + (0 if termine else gamma * Q[suivant, action_suiv])
            Q[etat, action] += alpha * (cible - Q[etat, action]); etat, action = suivant, action_suiv; total += recompense
            if termine or tronque: break
        gains.append(total)
    env.close(); return Q, np.array(gains)

Q, gains = entrainer_sarsa(GRAINE)
print('Politique SARSA (argmax par état):', Q.argmax(axis=1))
fleches = np.where(Q.max(axis=1) > 0, np.array(['←', '↓', '→', '↑'])[Q.argmax(axis=1)], '·')
print('Grille 4×4 :\n', fleches.reshape(4, 4))
print('Taux de réussite (politique gloutonne, graine 42):', evaluer_politique(Q))
taux = [evaluer_politique(entrainer_sarsa(s)[0]) for s in range(5)]
RL['SARSA'] = taux
print('SARSA sur 5 graines : taux de réussite =', taux, '| moyenne =', round(float(np.mean(taux)), 3))
if 'Q-learning' in RL:
    print('Comparaison (moyenne sur 5 graines) -> Q-learning:', round(float(np.mean(RL['Q-learning'])), 3), '| SARSA:', round(float(np.mean(RL['SARSA'])), 3))
````

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
recompenses = np.array([0., 1., -1.]); termine = np.array([False, True, False]); meilleur_Q_suivant = np.array([2., 3., 1.]); gamma = .95
cibles = recompenses + gamma * (~termine) * meilleur_Q_suivant     # y = r + gamma * (1 - d) * max Q_cible(s', a')
print('Cibles de Bellman:', cibles)
assert np.allclose(cibles, [1.9, 1., -.05])
print('Exemple du cours (r=1, gamma=0,9, d=0, maxQ=2) : y =', 1 + .9 * (1 - 0) * 2)
````
